```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("../data/processed/wind_20260601.csv")

df.head()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["Density"])
plt.title("Solar Wind Proton Density")
plt.xlabel("Time Index")
plt.ylabel("Density")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,5))
plt.plot(df["Vx"])
plt.title("Solar Wind Velocity Vx")
plt.xlabel("Time Index")
plt.ylabel("km/s")
plt.grid(True)
plt.show()
```
